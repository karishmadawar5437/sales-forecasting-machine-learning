# Walmart Sales Forecasting Project

## Notebook 05 – Feature Engineering

### Objective

Create new features from the daily sales dataset that will improve machine learning model performance for forecasting.

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_excel("../data/processed/daily_sales.xlsx")
```

```python
df.head()
```

```python
df.info()
```

```python
df.shape
```

```python
df.columns
```

```python
df.rename(columns={"index": "Order Date"}, inplace=True)
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])
```

```python
df["Sales"] = df["Sales"].fillna(0)
```

```python
df.isnull().sum()
```

```python
df["Year"] = df["Order Date"].dt.year

df["Quarter"] = df["Order Date"].dt.quarter

df["Month"] = df["Order Date"].dt.month

df["Week"] = df["Order Date"].dt.isocalendar().week.astype(int)

df["Day"] = df["Order Date"].dt.day

df["DayOfWeek"] = df["Order Date"].dt.dayofweek

df["DayName"] = df["Order Date"].dt.day_name()

df["IsWeekend"] = (df["DayOfWeek"] >= 5).astype(int)
```

```python
df["Lag_1"] = df["Sales"].shift(1)

df["Lag_7"] = df["Sales"].shift(7)

df["Lag_30"] = df["Sales"].shift(30)
```

```python
df["Rolling_Mean_7"] = df["Sales"].rolling(7).mean()

df["Rolling_Mean_30"] = df["Sales"].rolling(30).mean()

df["Rolling_STD_7"] = df["Sales"].rolling(7).std()
```

```python
df["Expanding_Mean"] = df["Sales"].expanding().mean()
```

```python
df["Month_Sin"] = np.sin(2*np.pi*df["Month"]/12)

df["Month_Cos"] = np.cos(2*np.pi*df["Month"]/12)

df["Day_Sin"] = np.sin(2*np.pi*df["DayOfWeek"]/7)

df["Day_Cos"] = np.cos(2*np.pi*df["DayOfWeek"]/7)
```

```python
df.isnull().sum()
```

```python
df = df.dropna().reset_index(drop=True)
```

```python
df.isnull().sum()
```

```python
df.head()
```

```python
df.info()
```

```python
df.shape
```

```python
df = df.drop(columns=["DayName"], errors="ignore")

df.to_excel("../data/processed/sales_feature_engineered.xlsx", index=False)

print("✅ sales_feature_engineered.xlsx saved successfully!")
```

# Business Insights

- Calendar features capture seasonality.
- Lag features provide historical sales information.
- Rolling statistics smooth short-term fluctuations.
- Expanding mean captures long-term sales trends.
- Cyclical features help machine learning models understand repeating time patterns.
- The dataset is now ready for model building.
